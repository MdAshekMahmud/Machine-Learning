```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = sns.load_dataset('tips')
data.head()
```

```python
sns.set_style(style="ticks")

tips = sns.load_dataset('tips')

sns.relplot(x='total_bill', y='tip', data=tips)
plt.show()
```

## Grouping data points on the basis of category, here as time.

```python
sns.set(style='ticks')

sns.relplot(x='total_bill',
            y='tip',
            hue='time',
            data=tips)
```

```python
sns.relplot(x='total_bill',
            y='tip',
            hue='day',
            col='time',
            row='sex',
            data=tips,
            )
```

Using size attribute, we can see data points having different size.

```python
sns.relplot(x='total_bill',
            y='tip',
            hue='day',
            size='size',
            data=tips)
```

```python
sns.relplot(x='total_bill',
            y='tip',
            kind='line',
            data=tips)
```

```python
markers = {'Lunch': 's', 'Dinner': 'X'}

ax = sns.scatterplot(x='total_bill',
                     y='tip',
                     style='time',
                     markers=markers,
                     data=tips)
```

```python
iris = sns.load_dataset('iris')

sns.scatterplot(x=iris.sepal_length,
                y=iris.sepal_width,
                hue=iris.species,
                style=iris.species,
                )
```

```python
sns.set(style='whitegrid')

fmri = sns.load_dataset('fmri')

sns.lineplot(x='timepoint',
             y='signal',
             data=fmri)
```

Groupint data points on the basis of category, here as region and event

```python
sns.lineplot(x='timepoint',
             y='signal',
             hue='region',
             style='event',
             data=fmri)
```

```python
dots = sns.load_dataset('dots').query("align == 'dots'")

sns.lineplot(x='time',
             y='firing_rate',
             hue='coherence',
             style='choice',
             data=dots)
```
